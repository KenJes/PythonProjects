## Sección 1 — Tipos de Delitos Informáticos

Vamos a usar un **diccionario de diccionarios** para organizar los principales tipos de delitos informáticos.  
Recuerda: en Python, un diccionario nos permite guardar información con una **clave** (nombre) y un **valor** (dato).

```python
# ============================================================
# DICCIONARIO DE DELITOS INFORMÁTICOS
# Cada clave es el nombre del delito.
# Cada valor es otro diccionario con: descripción, ejemplo y consecuencia.
# ============================================================

delitos_informaticos = {
    "acceso_no_autorizado": {
        "descripcion": "Entrar a un sistema, red o computadora sin permiso del dueño.",
        "ejemplo": "Adivinar la contraseña del correo de alguien para leer sus mensajes.",
        "consecuencia": "Hasta 2 años de prisión y multa (Art. 211 bis, CPF México)."
    },
    "phishing": {
        "descripcion": "Engañar a una persona haciéndose pasar por una entidad confiable para robar datos.",
        "ejemplo": "Un correo falso del banco que pide ingresar tu usuario y contraseña.",
        "consecuencia": "Fraude electrónico: hasta 9 años de prisión según el monto defraudado."
    },
    "fraude_electronico": {
        "descripcion": "Obtener dinero o beneficios de forma ilegal usando medios digitales.",
        "ejemplo": "Crear una tienda en línea falsa para cobrar productos que nunca se envían.",
        "consecuencia": "De 3 meses a 12 años de prisión según el valor del fraude (Art. 386, CPF)."
    },
    "malware": {
        "descripcion": "Software malicioso diseñado para dañar, espiar o tomar control de un sistema.",
        "ejemplo": "Un virus que cifra todos los archivos de una empresa y pide dinero para liberarlos (ransomware).",
        "consecuencia": "Daño informático: hasta 8 años de prisión (Art. 211 bis 1, CPF México)."
    },
    "robo_de_identidad": {
        "descripcion": "Usar los datos personales de otra persona sin su consentimiento.",
        "ejemplo": "Solicitar un crédito bancario usando el nombre, CURP y RFC de otra persona.",
        "consecuencia": "Hasta 5 años de prisión (Ley Federal de Protección de Datos Personales)."
    }
}

# ============================================================
# Ahora recorremos el diccionario para mostrar la información
# de forma ordenada. Usamos un ciclo 'for' sobre los elementos.
# ============================================================

print("=" * 60)
print("       CATÁLOGO DE DELITOS INFORMÁTICOS")
print("=" * 60)

# enumerate() nos da un número (i) y el par (clave, valor) al mismo tiempo
for i, (clave, info) in enumerate(delitos_informaticos.items(), start=1):
    # Reemplazamos los guiones bajos por espacios y ponemos en mayúsculas
    nombre_legible = clave.replace("_", " ").upper()
    print(f"\n{i}. {nombre_legible}")
    print(f"   📌 Descripción : {info['descripcion']}")
    print(f"   🔍 Ejemplo     : {info['ejemplo']}")
    print(f"   ⚖️  Consecuencia: {info['consecuencia']}")

print("\n" + "=" * 60)
```

## Sección 2 — Simulador de Detección de Phishing

El **phishing** es uno de los ataques más comunes. Los atacantes envían correos electrónicos falsos  
que imitan a bancos, tiendas o redes sociales para robar tus datos.

Vamos a construir un **analizador básico** que detecta señales de alerta en un mensaje usando **listas**.

```python
# ============================================================
# SIMULADOR DE DETECCIÓN DE PHISHING
# Usamos listas para guardar palabras y dominios sospechosos.
# ============================================================

# Lista de palabras que aparecen frecuentemente en correos de phishing
palabras_sospechosas = [
    "urgente", "gratis", "ganaste", "premio", "haz clic",
    "verifica ahora", "suspendido", "confirmación inmediata",
    "dinero fácil", "oferta exclusiva", "cuenta bloqueada",
    "actualiza tus datos", "gana dinero", "toca aquí"
]

# Lista de dominios de correo falsos o sospechosos
# (nótese cómo imitan dominios reales con pequeños cambios)
dominios_sospechosos = [
    "paypa1.com",       # Imita a paypal.com (cambia 'l' por '1')
    "amaz0n.com",       # Imita a amazon.com (cambia 'o' por '0')
    "bancomer-seguro.net",   # Dominio inventado que parece oficial
    "bbva-verificacion.org", # Los bancos reales no usan .org
    "soporte-microsoft.info",
    "netflix-cobro.com",
    "sat-devolucion.net"     # El SAT solo usa dominios .gob.mx
]


def analizar_correo(remitente, asunto, cuerpo):
    """
    Analiza un correo electrónico y detecta señales de phishing.
    Parámetros:
        remitente (str): Dirección de correo del emisor.
        asunto    (str): Asunto del correo.
        cuerpo    (str): Texto del correo.
    """
    print("\n" + "=" * 55)
    print("  ANALIZADOR DE CORREOS SOSPECHOSOS")
    print("=" * 55)
    print(f"  De     : {remitente}")
    print(f"  Asunto : {asunto}")
    print(f"  Cuerpo : {cuerpo[:80]}...")  # Solo mostramos los primeros 80 caracteres
    print("-" * 55)

    alertas = []  # Lista donde guardaremos las alertas encontradas

    # --- REVISIÓN 1: ¿El dominio del remitente es sospechoso? ---
    # Separamos el dominio del correo usando split('@')
    if "@" in remitente:
        dominio_remitente = remitente.split("@")[1].lower()
        if dominio_remitente in dominios_sospechosos:
            alertas.append(f"Dominio del remitente sospechoso: @{dominio_remitente}")

    # --- REVISIÓN 2: ¿El asunto contiene palabras sospechosas? ---
    asunto_lower = asunto.lower()
    for palabra in palabras_sospechosas:
        if palabra in asunto_lower:
            alertas.append(f"Palabra sospechosa en el asunto: '{palabra}'")

    # --- REVISIÓN 3: ¿El cuerpo contiene palabras sospechosas? ---
    cuerpo_lower = cuerpo.lower()
    for palabra in palabras_sospechosas:
        if palabra in cuerpo_lower:
            alertas.append(f"Palabra sospechosa en el cuerpo: '{palabra}'")

    # --- RESULTADO ---
    if len(alertas) == 0:
        print("  ✅ RESULTADO: El correo parece legítimo. No se detectaron alertas.")
    elif len(alertas) <= 2:
        print("  ⚠️  RESULTADO: PRECAUCIÓN — Se encontraron señales leves.")
        for alerta in alertas:
            print(f"     • {alerta}")
    else:
        print("  🚨 RESULTADO: PELIGRO — Este correo tiene muchas señales de PHISHING.")
        print("     NO hagas clic en ningún enlace ni compartas datos.")
        for alerta in alertas:
            print(f"     • {alerta}")

    print(f"\n  Total de alertas detectadas: {len(alertas)}")
    print("=" * 55)


# ============================================================
# CASO 1: Correo de phishing bancario (PELIGROSO)
# ============================================================
analizar_correo(
    remitente="seguridad@bbva-verificacion.org",
    asunto="URGENTE: Tu cuenta ha sido suspendida",
    cuerpo="Estimado cliente, tu cuenta bancaria ha sido bloqueada por actividad inusual. "
           "Verifica ahora tus datos haciendo clic en el siguiente enlace para evitar "
           "la cancelación definitiva de tu cuenta. Confirmación inmediata requerida."
)

# ============================================================
# CASO 2: Correo legítimo de un profesor
# ============================================================
analizar_correo(
    remitente="profesor.garcia@universidad.edu.mx",
    asunto="Material de la clase del viernes",
    cuerpo="Hola estudiantes, les comparto el material de apoyo para la práctica "
           "del viernes. Está disponible en la plataforma institucional. Saludos."
)

# ============================================================
# CASO 3: Supuesta notificación de premio (phishing)
# ============================================================
analizar_correo(
    remitente="premios@amaz0n.com",
    asunto="¡Ganaste un premio! Oferta exclusiva solo para ti — gratis",
    cuerpo="Felicidades, eres el ganador de nuestra rifa especial. "
           "Toca aquí para reclamar tu premio gratis. Dinero fácil garantizado. "
           "Actualiza tus datos para recibir tu transferencia."
)
```

## Sección 3 — Evaluador de Contraseñas: Débiles vs. Fuertes

Una **contraseña débil** es la primera puerta de entrada para un atacante.  
Los atacantes usan **ataques de diccionario**: prueban miles de contraseñas comunes automáticamente  
hasta encontrar la correcta. Por eso es vital usar contraseñas fuertes y únicas.

```python
# ============================================================
# EVALUADOR DE CONTRASEÑAS
# Usamos listas para guardar contraseñas comunes (lista negra)
# y strings para analizar la contraseña del usuario.
# ============================================================

# Las 20 contraseñas más usadas en el mundo (y las más inseguras)
# Fuente: NordPass Annual Report 2024
contrasenas_comunes = [
    "123456", "password", "123456789", "12345678", "12345",
    "1234567", "1234567890", "qwerty", "abc123", "111111",
    "1234", "iloveyou", "admin", "letmein", "monkey",
    "1q2w3e4r", "master", "dragon", "pass", "superman",
    "batman", "123123", "000000", "contraseña", "mexico123"
]

# Lista de caracteres especiales que aumentan la seguridad
caracteres_especiales = ["!", "@", "#", "$", "%", "^", "&", "*", "-", "_", "+", "="]


def evaluar_contrasena(contrasena):
    """
    Evalúa la fortaleza de una contraseña.
    Criterios: longitud, mayúsculas, números, caracteres especiales, lista negra.
    """
    print("\n" + "-" * 50)
    print(f"  Analizando contraseña: {'*' * len(contrasena)}  ({len(contrasena)} caracteres)")
    print("-" * 50)

    puntos = 0        # Puntaje acumulado (máximo 5)
    problemas = []    # Lista de problemas encontrados
    aciertos = []     # Lista de cosas que hace bien

    # --- CRITERIO 1: Longitud mínima de 8 caracteres ---
    if len(contrasena) >= 8:
        puntos += 1
        aciertos.append("Tiene 8 o más caracteres")
    else:
        problemas.append("Muy corta (mínimo 8 caracteres)")

    # --- CRITERIO 2: Longitud recomendada de 12+ caracteres ---
    if len(contrasena) >= 12:
        puntos += 1
        aciertos.append("Tiene 12 o más caracteres (excelente)")

    # --- CRITERIO 3: Contiene al menos una letra mayúscula ---
    # Recorremos cada carácter y verificamos si alguno es mayúscula
    tiene_mayuscula = False
    for caracter in contrasena:
        if caracter.isupper():
            tiene_mayuscula = True
    if tiene_mayuscula:
        puntos += 1
        aciertos.append("Contiene letras mayúsculas")
    else:
        problemas.append("No contiene mayúsculas")

    # --- CRITERIO 4: Contiene al menos un número ---
    tiene_numero = False
    for caracter in contrasena:
        if caracter.isdigit():
            tiene_numero = True
    if tiene_numero:
        puntos += 1
        aciertos.append("Contiene números")
    else:
        problemas.append("No contiene números")

    # --- CRITERIO 5: Contiene al menos un carácter especial ---
    tiene_especial = False
    for caracter in contrasena:
        if caracter in caracteres_especiales:
            tiene_especial = True
    if tiene_especial:
        puntos += 1
        aciertos.append("Contiene caracteres especiales (!@#$...)")
    else:
        problemas.append("No contiene caracteres especiales")

    # --- CRITERIO 6: No está en la lista de contraseñas comunes ---
    if contrasena.lower() in contrasenas_comunes:
        puntos = 0  # Anulamos todos los puntos — es una contraseña pública
        problemas.append("¡Está en la lista de las contraseñas MÁS USADAS DEL MUNDO!")

    # --- RESULTADO FINAL ---
    print("  ✅ Puntos positivos:")
    if aciertos:
        for a in aciertos:
            print(f"     + {a}")
    else:
        print("     (ninguno)")

    print("  ❌ Problemas:")
    if problemas:
        for p in problemas:
            print(f"     - {p}")
    else:
        print("     (ninguno)")

    # Determinamos el nivel según los puntos obtenidos
    if puntos == 0:
        nivel = "🔴 MUY DÉBIL — Un atacante la adivinaría en segundos"
    elif puntos <= 2:
        nivel = "🟠 DÉBIL — Vulnerable a ataques de diccionario"
    elif puntos == 3:
        nivel = "🟡 MODERADA — Puede mejorarse"
    elif puntos == 4:
        nivel = "🟢 FUERTE — Buena contraseña"
    else:
        nivel = "🟢✅ MUY FUERTE — Excelente contraseña"

    print(f"\n  Puntaje: {puntos}/5")
    print(f"  Nivel  : {nivel}")


# ============================================================
# Probamos con distintas contraseñas
# ============================================================
print("══════════════════════════════════════════════════")
print("        EVALUADOR DE CONTRASEÑAS SEGURAS")
print("══════════════════════════════════════════════════")

evaluar_contrasena("123456")              # Muy débil, en lista negra
evaluar_contrasena("miperro2024")         # Débil, sin mayúsculas ni especiales
evaluar_contrasena("MiPerro2024")         # Moderada
evaluar_contrasena("MiPerro!2024#Fluffy") # Muy fuerte
```

## Sección 4 — Cifrado César: Cómo se Oculta Información

Los **criminales digitales** cifran sus comunicaciones para que no puedan ser leídas si son interceptadas.  
Pero el cifrado también es la base de la **seguridad legítima**: bancos, WhatsApp y webs HTTPS lo usan.

El **Cifrado César** es el más antiguo y simple: desplaza cada letra del alfabeto un número fijo de posiciones.  
Ejemplo con desplazamiento 3: `A → D`, `B → E`, `HOLA → KROD`

Lo implementaremos usando **listas** para que veas el proceso paso a paso.

```python
# ============================================================
# CIFRADO CÉSAR CON LISTAS
# Usamos una lista con todas las letras del alfabeto.
# El cifrado desplaza cada letra n posiciones hacia adelante.
# ============================================================

# Lista del alfabeto en español (sin ñ para simplificar)
alfabeto = [
    'a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm',
    'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z'
]
# len(alfabeto) = 26  →  los índices van de 0 a 25


def cifrar_cesar(mensaje, desplazamiento):
    """
    Cifra un mensaje usando el cifrado César.
    Cada letra se desplaza 'desplazamiento' posiciones en el alfabeto.
    Los espacios y caracteres no alfabéticos se conservan sin cambio.
    """
    mensaje_cifrado = []  # Lista donde guardaremos cada carácter cifrado

    for caracter in mensaje.lower():  # Convertimos todo a minúsculas
        if caracter in alfabeto:
            # Encontramos la posición original en el alfabeto
            posicion_original = alfabeto.index(caracter)

            # Calculamos la nueva posición con el desplazamiento
            # El operador % 26 asegura que 'volvemos al inicio' si pasamos la 'z'
            # Ejemplo: 'z' está en posición 25. Con desplazamiento 3: (25+3) % 26 = 2 → 'c'
            nueva_posicion = (posicion_original + desplazamiento) % 26

            # Tomamos la letra en la nueva posición
            caracter_cifrado = alfabeto[nueva_posicion]
            mensaje_cifrado.append(caracter_cifrado)
        else:
            # Espacios, números y signos se conservan tal cual
            mensaje_cifrado.append(caracter)

    # Unimos todos los caracteres de la lista en un solo string
    return "".join(mensaje_cifrado)


def descifrar_cesar(mensaje_cifrado, desplazamiento):
    """
    Descifra un mensaje cifrado con César.
    Para descifrar, simplemente ciframos en la dirección opuesta
    (restamos el desplazamiento en vez de sumarlo).
    """
    # Desplazamiento inverso: 26 - desplazamiento
    # Ejemplo: si ciframos con 3, desciframos con 26-3=23
    return cifrar_cesar(mensaje_cifrado, 26 - desplazamiento)


# ============================================================
# DEMOSTRACIÓN PASO A PASO
# ============================================================

print("=" * 55)
print("        CIFRADO CÉSAR — DEMOSTRACIÓN")
print("=" * 55)

mensaje_original = "transferencia a las diez de la noche"
clave = 7  # Desplazamiento secreto (solo el emisor y receptor lo conocen)

# Paso 1: Cifrar
mensaje_cifrado = cifrar_cesar(mensaje_original, clave)
# Paso 2: Descifrar
mensaje_descifrado = descifrar_cesar(mensaje_cifrado, clave)

print(f"\n  Desplazamiento (clave secreta): {clave}")
print(f"\n  Mensaje ORIGINAL  : {mensaje_original}")
print(f"  Mensaje CIFRADO   : {mensaje_cifrado}")
print(f"  Mensaje DESCIFRADO: {mensaje_descifrado}")

print("\n" + "-" * 55)
print("  Visualización del desplazamiento:")
print(f"  Alfabeto normal  : {' '.join(alfabeto)}")
alfabeto_desplazado = alfabeto[clave:] + alfabeto[:clave]
print(f"  Alfabeto + {clave} pos.: {' '.join(alfabeto_desplazado)}")

# ============================================================
# ¿Por qué el cifrado César es INSEGURO hoy en día?
# Podemos 'romperlo' probando los 25 posibles desplazamientos
# ============================================================

print("\n" + "=" * 55)
print("  ATAQUE DE FUERZA BRUTA al Cifrado César")
print("  (probamos los 25 desplazamientos posibles)")
print("=" * 55)
print(f"  Mensaje interceptado: '{mensaje_cifrado}'\n")

for intento in range(1, 26):
    resultado = descifrar_cesar(mensaje_cifrado, intento)
    print(f"  Desplazamiento {intento:2d}: {resultado}")
```

### 🧠 Reflexión 4
- En el ataque de fuerza bruta, ¿pudiste identificar el mensaje original entre los 25 intentos?
- El Cifrado César tiene solo **25 combinaciones posibles** — una computadora lo rompe en microsegundos.
- Los algoritmos modernos como **AES-256** tienen $2^{256}$ combinaciones posibles —  
  ni todas las computadoras del mundo juntas podrían romperlo por fuerza bruta en millones de años.
- ¿Para qué sirve entonces estudiar el Cifrado César? Es la base conceptual de criptografía más avanzada.
