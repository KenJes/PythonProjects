# 1. ¿Qué es la Criptografía?

<div style="background:#dce9ff; border-left: 6px solid #2f57d9; padding: 16px 20px; border-radius: 8px; margin-bottom: 16px; color:#10243e; box-shadow:0 1px 3px rgba(15,23,42,0.08);">
<b>Definición:</b> La <em>criptografía</em> (del griego <em>kryptós</em> = oculto, <em>graphé</em> = escritura) es la ciencia de proteger información mediante técnicas que la transforman en un formato ilegible para quienes no tienen autorización.
</div>

## Vocabulario esencial

| Término | Descripción |
|---|---|
| **Texto claro** (*plaintext*) | Mensaje original, legible |
| **Texto cifrado** (*ciphertext*) | Mensaje transformado, ilegible |
| **Cifrado** (*cipher*) | Algoritmo de transformación |
| **Clave** (*key*) | Parámetro secreto que controla el cifrado |
| **Cifrar** (*encrypt*) | Convertir texto claro → cifrado |
| **Descifrar** (*decrypt*) | Convertir texto cifrado → claro |
| **Criptoanálisis** | Arte de romper cifrados sin conocer la clave |

## Criptografía ≠ Seguridad total

```
Criptografía   →   herramienta fundamental
Seguridad      →   sistema completo (personas + procesos + tecnología)
```

> La criptografía protege los **datos**; no protege contra malas contraseñas, ingeniería social ni vulnerabilidades de software.

## Principios del buen diseño criptográfico

### Principio de Kerckhoffs (1883)
> *"Un sistema criptográfico debe ser seguro incluso si todo sobre el sistema, excepto la clave, es de conocimiento público."*

→ La **seguridad reside en la clave**, no en el secreto del algoritmo.

### Ley de Schneier
> *"Cualquiera puede diseñar un algoritmo de cifrado que él mismo no pueda romper. No es difícil diseñar algo que ni siquiera tú puedas atacar."*

→ Solo los algoritmos **públicamente revisados y probados** merecen confianza.

### Confusión y Difusión (Shannon, 1949)
| Propiedad | Significado |
|---|---|
| **Confusión** | Cada bit del cifrado depende de múltiples partes de la clave |
| **Difusión** | Cambiar 1 bit del texto claro cambia ~50% del texto cifrado |

# 4. Historia: De César a Turing

## Línea del tiempo

```
 ~1900 a.C.  →  Egipto: jeroglíficos no estándar (1.ª criptografía conocida)
 ~600 a.C.   →  Escítala espartana (transposición con cilindro)
  50 a.C.    →  Cifrado César (sustitución por desplazamiento)
  830 d.C.   →  Al-Kindi: primer criptoanálisis (análisis de frecuencia)
 1467        →  Leon Battista Alberti: disco cifrador (polialfabético)
 1553        →  Cifrado de Vigenère (múltiples alfabetos de sustitución)
 1854        →  Cifrado Playfair (sustitución por pares de letras)
 1917        →  One-Time Pad (Vernam): matemáticamente perfecto
 1918-1945   →  Enigma alemana → Alan Turing la rompe en Bletchley Park
 1977        →  DES (Data Encryption Standard) - primer estándar moderno
 1978        →  RSA: criptografía asimétrica pública
 2001        →  AES reemplaza a DES como estándar mundial
 2009        →  Bitcoin: criptografía para monedas digitales
 2016+       →  NIST inicia estandarización de criptografía post-cuántica
```

## Cifrado César
Cada letra se desplaza `n` posiciones en el alfabeto.

```
Texto claro:  H  O  L  A
Desplaz. +3:  K  R  Ñ  D   (en alfabeto español)
              ↑  ↑  ↑  ↑
              K  R  O  D   (en alfabeto inglés)
```

## La Máquina Enigma y Turing
- Enigma usaba **rotores mecánicos** que cambiaban el cifrado con cada pulsación
- Con 3 rotores: **158 quintillones** de configuraciones posibles
- Turing construyó la **Bomba** (Bombe): máquina electromecánica que explotaba patrones del lenguaje y errores operativos alemanes
- Estimación: **acortó la 2ª Guerra Mundial 2-4 años**

````python
# ============================================================
#  DEMO: Cifrado César
# ============================================================

def cifrar_cesar(texto: str, desplazamiento: int) -> str:
    """Cifra un texto usando el cifrado César."""
    resultado = []
    for char in texto.upper():
        if char.isalpha() and char.isascii():
            cifrado = chr((ord(char) - ord('A') + desplazamiento) % 26 + ord('A'))
            resultado.append(cifrado)
        else:
            resultado.append(char)
    return ''.join(resultado)

def descifrar_cesar(texto: str, desplazamiento: int) -> str:
    return cifrar_cesar(texto, -desplazamiento)

def fuerza_bruta_cesar(cifrado: str):
    """Intenta todos los desplazamientos posibles."""
    print(f"\nFuerza bruta sobre: '{cifrado}'")
    print("-" * 50)
    for k in range(26):
        candidato = descifrar_cesar(cifrado, k)
        print(f"  Clave {k:2d}: {candidato}")

# --- Demo ---
mensaje = "HOLA MUNDO"
clave = 13  # ROT-13: caso especial simétrico
cifrado = cifrar_cesar(mensaje, clave)
descifrado = descifrar_cesar(cifrado, clave)

print("=" * 50)
print(" Cifrado César - Demostración")
print("=" * 50)
print(f"  Texto claro   : {mensaje}")
print(f"  Clave (ROT)   : {clave}")
print(f"  Texto cifrado : {cifrado}")
print(f"  Descifrado    : {descifrado}")
print()
print("Con ROT-13: cifrar dos veces devuelve el original")
print(f"  Cifrar '{cifrado}' -> {cifrar_cesar(cifrado, 13)}")

# Fuerza bruta
fuerza_bruta_cesar("KHOOR ZRUOG")
````

````python
# ============================================================
#  Análisis de Frecuencia (Al-Kindi, ~830 d.C.)
# ============================================================
import collections
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

texto_ejemplo = (
    "EN UN LUGAR DE LA MANCHA DE CUYO NOMBRE NO QUIERO ACORDARME "
    "NO HA MUCHO TIEMPO QUE VIVIA UN HIDALGO DE LOS DE LANZA EN ASTILLERO "
    "ADARGA ANTIGUA ROCIN FLACO Y GALGO CORREDOR"
 )

clave_demo = 7
cifrado_demo = cifrar_cesar(texto_ejemplo, clave_demo)

def contar_letras(texto):
    return collections.Counter(c for c in texto if c.isalpha())

freq_original = contar_letras(texto_ejemplo)
freq_cifrado = contar_letras(cifrado_demo)

letras = sorted(freq_original.keys())
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
fig.suptitle("Análisis de Frecuencia: Por qué el Cifrado César es débil", fontsize=13, fontweight='bold')

for ax, freq, titulo, color in [
    (axes[0], freq_original, "Texto Original", "#3498db"),
    (axes[1], freq_cifrado, f"Texto Cifrado (clave={clave_demo})", "#e74c3c")
]:
    sorted_freq = sorted(freq.items())
    ax.bar([x[0] for x in sorted_freq], [x[1] for x in sorted_freq], color=color, alpha=0.8)
    ax.set_title(titulo)
    ax.set_xlabel("Letra")
    ax.set_ylabel("Frecuencia")
    ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
print("\nEl patrón de frecuencias se conserva: el análisis de frecuencia rompe el cifrado.")
print(f"La letra más frecuente en el cifrado es la misma que '{max(freq_original, key=freq_original.get)}' desplazada {clave_demo} posiciones.")
````

# 5. Criptografía Simétrica

<div style="background:#fff0c7; border-left: 6px solid #c57b12; padding: 14px 18px; border-radius: 8px; margin-bottom: 16px; color:#4a3310; box-shadow:0 1px 3px rgba(15,23,42,0.08);">
<b>Definición:</b> Emisor y receptor comparten la <em>misma clave</em> para cifrar y descifrar.
</div>

```
  ALICIA                              BOB
  Texto claro                         Texto claro
      │                                   ▲
      │ CIFRAR con clave K                │ DESCIFRAR con clave K
      ▼                                   │
  Texto cifrado ──── canal inseguro ──────┘
  
  Pregunta central: ¿cómo se comparte K de forma segura?
```

## Algoritmos principales

| Algoritmo | Año | Tamaño de bloque | Longitud de clave | Estado |
|---|---|---|---|---|
| **DES** | 1977 | 64 bits | 56 bits | Obsoleto (roto en 1999) |
| **3DES** | 1998 | 64 bits | 112/168 bits | Deprecado |
| **AES-128** | 2001 | 128 bits | 128 bits | Seguro |
| **AES-256** | 2001 | 128 bits | 256 bits | Estándar mundial |
| **ChaCha20** | 2008 | Stream | 256 bits | Usado en TLS 1.3 |

## Modos de operación AES

| Modo | Nombre | Característica |
|---|---|---|
| **ECB** | Electronic Codebook | Inseguro: bloques iguales generan cifrado igual |
| **CBC** | Cipher Block Chaining | Cada bloque depende del anterior + IV |
| **CTR** | Counter Mode | Convierte bloque-cipher en stream cipher |
| **GCM** | Galois/Counter Mode | Cifrado + autenticación (AEAD) |

## Ventajas y limitaciones
| Ventajas | Limitaciones |
|---|---|
| Muy rápido (hardware dedicado AES-NI) | Problema de distribución de claves |
| Adecuado para grandes volúmenes de datos | N participantes necesitan N(N-1)/2 claves |
| AES-256 resiste mejor ataques cuánticos conocidos | No provee autenticación por sí solo |

````python
# ============================================================
#  DEMO: AES-256-GCM (Cifrado Autenticado)
# ============================================================
import os
import time
try:
    from cryptography.hazmat.primitives.ciphers.aead import AESGCM
    CRYPTO_OK = True
except ImportError:
    CRYPTO_OK = False
    print("Instala 'cryptography' con: pip install cryptography")

if CRYPTO_OK:
    # Generar clave de 256 bits (32 bytes)
    clave_aes = AESGCM.generate_key(bit_length=256)
    aesgcm = AESGCM(clave_aes)

    # Nonce: número de uso único (96 bits para GCM)
    nonce = os.urandom(12)
    mensaje = b"Transferencia: $10,000 a cuenta 1234-5678"

    # Datos adicionales autenticados (AAD) - no se cifran, sí se autentican
    aad = b"usuario:alice|timestamp:20260527"

    # Cifrar
    t0 = time.perf_counter()
    cifrado_aes = aesgcm.encrypt(nonce, mensaje, aad)
    t_enc = (time.perf_counter() - t0) * 1e6

    # Descifrar
    t0 = time.perf_counter()
    descifrado_aes = aesgcm.decrypt(nonce, cifrado_aes, aad)
    t_dec = (time.perf_counter() - t0) * 1e6

    print("=" * 60)
    print(" AES-256-GCM - Cifrado autenticado")
    print("=" * 60)
    print(f"  Clave (hex)      : {clave_aes.hex()[:32]}...")
    print(f"  Nonce (hex)      : {nonce.hex()}")
    print(f"  Mensaje original : {mensaje.decode()}")
    print(f"  Cifrado (hex)    : {cifrado_aes.hex()[:48]}...")
    print(f"  Tamaño cifrado   : {len(cifrado_aes)} bytes (+16 bytes de tag GCM)")
    print(f"  Descifrado       : {descifrado_aes.decode()}")
    print(f"  Tiempo cifrado   : {t_enc:.1f} µs")
    print(f"  Tiempo descifrado: {t_dec:.1f} µs")

    # Demostrar que la autenticación falla si se modifica el cifrado
    print("\n--- Prueba de integridad (bit-flip attack) ---")
    cifrado_modificado = bytearray(cifrado_aes)
    cifrado_modificado[0] ^= 0xFF  # Invertir bits del primer byte
    try:
        aesgcm.decrypt(nonce, bytes(cifrado_modificado), aad)
    except Exception as e:
        print(f"  Modificación detectada: {type(e).__name__}")
        print("  El tag GCM rechazó el mensaje alterado.")

    # Demostrar problema del modo ECB
    print("\n--- ¿Por qué ECB es inseguro? ---")
    from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
    from cryptography.hazmat.backends import default_backend
    clave_ecb = os.urandom(32)
    bloque1 = b"TRANSFERIR $100 "  # 16 bytes
    bloque2 = b"TRANSFERIR $100 "  # Mismo contenido
    for bloque in [bloque1, bloque2]:
        c = Cipher(algorithms.AES(clave_ecb), modes.ECB(), backend=default_backend())
        enc = c.encryptor()
        print(f"  '{bloque.decode()}' -> {enc.update(bloque).hex()}")
    print("  Bloques idénticos producen cifrados idénticos: el patrón queda visible.")
````

# 6. Criptografía Asimétrica (Clave Pública)

<div style="background:#dcecff; border-left: 6px solid #1769aa; padding: 14px 18px; border-radius: 8px; margin-bottom: 16px; color:#0f2740; box-shadow:0 1px 3px rgba(15,23,42,0.08);">
<b>Revolución de 1976 (Diffie & Hellman):</b> Cada usuario tiene <em>dos claves matemáticamente relacionadas</em>: una <b>pública</b> (puede compartirse) y una <b>privada</b> (secreta). Lo que cifra una, solo lo descifra la otra.
</div>

## ¿Cómo funciona?

```
  BOB genera su par de claves:
    Clave pública  -> la publica en internet
    Clave privada  -> la guarda en secreto

  ALICIA quiere enviar mensaje secreto a BOB:
    1. Obtiene la clave pública de Bob
    2. Cifra el mensaje con ella
    3. Solo Bob, con su clave privada, puede descifrar

  BOB quiere firmar un documento:
    1. Genera la firma con su clave privada
    2. Cualquiera verifica con su clave pública
```

## Algoritmos principales

| Algoritmo | Base matemática | Uso típico | Seguridad |
|---|---|---|---|
| **RSA** | Factorización de enteros grandes | Cifrado, firma | 2048+ bits |
| **DSA** | Logaritmo discreto | Firma digital | Solo firma |
| **ECDSA** | Curvas elípticas | Firma (Bitcoin, TLS) | 256 bits ≈ RSA-3072 |
| **ECDH** | Curvas elípticas | Intercambio de claves | TLS 1.3 |
| **Ed25519** | Curva de Edwards | Firma rápida | Recomendado |

## RSA: La magia matemática

$$n = p \times q \quad (p, q \text{ primos grandes})$$
$$e \cdot d \equiv 1 \pmod{\phi(n)}$$
$$\text{Cifrado: } c = m^e \bmod n$$
$$\text{Descifrado: } m = c^d \bmod n$$

**Clave pública:** $(n, e)$ — **Clave privada:** $(n, d)$

La seguridad depende de que **factorizar $n$** en sus primos $p$ y $q$ sea computacionalmente inviable.

## Intercambio de claves Diffie-Hellman

```
  Color base público: amarillo
  
  ALICIA elige rojo (secreto) -> mezcla -> naranja (envía a Bob)
  BOB elige azul (secreto) -> mezcla -> verde (envía a Alice)
  
  ALICIA: verde + rojo = marrón (clave compartida)
  BOB:    naranja + azul = marrón (misma clave)
  EVA:    Solo ve naranja y verde; no puede reconstruir el secreto
```

````python
# ============================================================
#  DEMO: RSA - Cifrado y Firma Digital
# ============================================================
import time
try:
    from cryptography.hazmat.primitives.asymmetric import rsa, padding
    from cryptography.hazmat.primitives import hashes, serialization
    from cryptography.hazmat.backends import default_backend
    CRYPTO_OK = True
except ImportError:
    CRYPTO_OK = False
    print("Instala 'cryptography' con: pip install cryptography")

if CRYPTO_OK:
    print("=" * 60)
    print(" RSA-2048 - Generando par de claves...")
    print("=" * 60)

    t0 = time.perf_counter()
    clave_privada = rsa.generate_private_key(
        public_exponent=65537,
        key_size=2048,
        backend=default_backend()
    )
    clave_publica = clave_privada.public_key()
    t_gen = time.perf_counter() - t0
    print(f"  Tiempo de generación de claves: {t_gen*1000:.1f} ms")

    # Mostrar tamaño de la clave
    pub_bytes = clave_publica.public_bytes(
        serialization.Encoding.PEM,
        serialization.PublicFormat.SubjectPublicKeyInfo
    )
    print("  Clave pública (PEM):")
    print("  " + pub_bytes.decode().replace('\n', '\n  ')[:200] + "...")

    # --- Cifrado asimétrico ---
    print("\n--- Cifrado con clave pública ---")
    mensaje_rsa = b"Clave de sesion AES: secret123"
    cifrado_rsa = clave_publica.encrypt(
        mensaje_rsa,
        padding.OAEP(
            mgf=padding.MGF1(algorithm=hashes.SHA256()),
            algorithm=hashes.SHA256(),
            label=None
        )
    )
    descifrado_rsa = clave_privada.decrypt(
        cifrado_rsa,
        padding.OAEP(
            mgf=padding.MGF1(algorithm=hashes.SHA256()),
            algorithm=hashes.SHA256(),
            label=None
        )
    )
    print(f"  Mensaje    : {mensaje_rsa.decode()}")
    print(f"  Cifrado    : {cifrado_rsa.hex()[:48]}...")
    print(f"  Descifrado : {descifrado_rsa.decode()}")

    # --- Firma digital ---
    print("\n--- Firma digital (autenticación + no repudio) ---")
    contrato = b"Contrato: pago de $50,000 el 27/05/2026"
    firma = clave_privada.sign(
        contrato,
        padding.PSS(
            mgf=padding.MGF1(hashes.SHA256()),
            salt_length=padding.PSS.MAX_LENGTH
        ),
        hashes.SHA256()
    )
    print(f"  Documento  : {contrato.decode()}")
    print(f"  Firma (hex): {firma.hex()[:48]}...")
    try:
        clave_publica.verify(
            firma, contrato,
            padding.PSS(mgf=padding.MGF1(hashes.SHA256()), salt_length=padding.PSS.MAX_LENGTH),
            hashes.SHA256()
        )
        print("  Firma verificada: el documento es auténtico e íntegro")
    except Exception:
        print("  Firma inválida")

    print("\nComparación de velocidad: RSA vs AES")
    print("  RSA (cifrar 30 bytes): ~1-5 ms")
    print("  AES (cifrar 1 GB)   : ~0.5 s")
    print("  En la práctica: RSA cifra la clave AES y AES cifra los datos.")
````

# 7. Funciones Hash e Integridad

<div style="background:#dcf6e6; border-left: 6px solid #208a55; padding: 14px 18px; border-radius: 8px; margin-bottom: 16px; color:#163525; box-shadow:0 1px 3px rgba(15,23,42,0.08);">
<b>Definición:</b> Una función hash criptográfica convierte datos de cualquier tamaño en una cadena de longitud fija (el <em>digest</em> o <em>huella digital</em>) de forma <em>determinista pero irreversible</em>.
</div>

## Propiedades matemáticas requeridas

| Propiedad | Descripción |
|---|---|
| **Determinismo** | `hash(X)` siempre produce el mismo resultado |
| **Eficiencia** | Rápido de calcular en cualquier dirección |
| **Resistencia a preimagen** | Dado `h`, imposible encontrar `X` tal que `hash(X)=h` |
| **Resistencia a segunda preimagen** | Dado `X`, imposible encontrar `Y≠X` con `hash(X)=hash(Y)` |
| **Resistencia a colisiones** | Imposible encontrar `X≠Y` tal que `hash(X)=hash(Y)` |
| **Efecto avalancha** | Cambiar 1 bit del input cambia ~50% del output |

## Algoritmos de hash

| Algoritmo | Tamaño digest | Estado | Uso |
|---|---|---|---|
| **MD5** | 128 bits | Roto (colisiones en 2004) | Solo checksums no críticos |
| **SHA-1** | 160 bits | Roto (2017, Google) | Evitar |
| **SHA-256** | 256 bits | Seguro | TLS, Bitcoin, Git |
| **SHA-3 (Keccak)** | Variable | Seguro | Estándar NIST 2015 |
| **BLAKE3** | Variable | Muy rápido | Moderno, recomendado |

## Aplicaciones del hash

```
1. Almacenamiento de contraseñas:
   bd almacena: bcrypt("miContraseña") -> no la contraseña en texto claro

2. Verificación de archivos:
   SHA256(archivo.iso) -> huella única para detectar corrupción

3. Firma digital:
   firmar(hash(documento)) en vez de firmar todo el documento

4. Blockchain:
   cada bloque contiene hash del bloque anterior -> cadena inmutable

5. Git:
   cada commit se identifica por el SHA-1/SHA-256 de su contenido
```

````python
# ============================================================
#  DEMO: Funciones Hash y Almacenamiento de Contraseñas
# ============================================================
import hashlib
import hmac
import os

print("=" * 60)
print(" Funciones Hash - Demostración")
print("=" * 60)

# --- Efecto avalancha ---
print("\n1. Efecto avalancha")
msg1 = "Hola, mundo!"
msg2 = "Hola, mundo."  # Solo cambia ! por .
h1 = hashlib.sha256(msg1.encode()).hexdigest()
h2 = hashlib.sha256(msg2.encode()).hexdigest()

bits_distintos = sum(bin(int(a, 16) ^ int(b, 16)).count('1')
                     for a, b in zip(h1, h2))
total_bits = 256

print(f"  Mensaje 1: '{msg1}' -> SHA256: {h1[:20]}...")
print(f"  Mensaje 2: '{msg2}' -> SHA256: {h2[:20]}...")
print(f"  Diferencia: {bits_distintos}/{total_bits} bits distintos ({bits_distintos/total_bits*100:.1f}%)")

# --- Múltiples algoritmos ---
print("\n2. Mismo texto, diferentes algoritmos")
texto = b"Criptografia 2026"
for algoritmo in ['md5', 'sha1', 'sha256', 'sha512', 'sha3_256']:
    h = hashlib.new(algoritmo, texto).hexdigest()
    print(f"  {algoritmo.upper():12s}: {h[:48]}... ({len(h)*4} bits)")

# --- Contraseñas: PBKDF2 vs texto claro ---
print("\n3. Almacenamiento seguro de contraseñas (PBKDF2-HMAC-SHA256)")
password = "MiContraseña123!"
salt = os.urandom(16)  # Sal aleatoria por usuario

dk = hashlib.pbkdf2_hmac('sha256', password.encode(), salt, 100_000)
print(f"  Contraseña : {password}")
print(f"  Salt (hex) : {salt.hex()}")
print(f"  Hash BD    : {dk.hex()}")
print("  Iteraciones: 100,000 (hace que la fuerza bruta sea mucho más lenta)")

# --- HMAC ---
print("\n4. HMAC - Hash con autenticación (MAC)")
clave_secreta = b"clave_servidor_secreta"
mensaje_api = b"user=alice&action=transfer&amount=100"
mac = hmac.new(clave_secreta, mensaje_api, hashlib.sha256).hexdigest()
print(f"  Mensaje  : {mensaje_api.decode()}")
print(f"  HMAC-256 : {mac}")
print("  Solo quien conoce la clave secreta puede generar o verificar este MAC.")
````

# 8. Nuevos Horizontes y Amenazas

## A. Esteganografía: Ocultar, no solo cifrar

<div style="background:#f7e1ef; border-left: 6px solid #8b3ea7; padding: 14px 18px; border-radius: 8px; margin-bottom: 12px; color:#39163d; box-shadow:0 1px 3px rgba(15,23,42,0.08);">
<b>Esteganografía</b> (del griego <em>steganos</em> = cubierto): arte de ocultar un mensaje <em>dentro de otro mensaje aparentemente inocente</em>, sin que nadie sospeche que existe información oculta.
</div>

**Criptografía vs Esteganografía:**
| | Criptografía | Esteganografía |
|---|---|---|
| **Objetivo** | Hacer ilegible el mensaje | Ocultar la existencia del mensaje |
| **Si te detectan** | Saben que hay mensaje cifrado | No saben que hay mensaje |
| **Mejor uso** | Juntos: mensaje cifrado + oculto | |

**Técnicas modernas:**
- **LSB (Least Significant Bit):** modifica el bit menos significativo de cada píxel de una imagen
- **Audio steganography:** oculta datos en frecuencias inaudibles
- **Text steganography:** espacios invisibles (Unicode) entre caracteres
- **DeepFake/AI:** canales encubiertos en redes neuronales

---
## B. Computación Cuántica: ¿El fin de la criptografía actual?

### Algoritmos cuánticos que rompen la criptografía actual

| Algoritmo cuántico | Problema resuelto | Criptografía vulnerable |
|---|---|---|
| **Shor (1994)** | Factorización en tiempo polinomial | RSA, DSA, ECC |
| **Grover (1996)** | Búsqueda en $\sqrt{N}$ | AES-128 y SHA-256 se debilitan |

### Criptografía Post-Cuántica (PQC) - NIST 2024

| Estándar NIST | Basado en | Uso |
|---|---|---|
| **ML-KEM** (CRYSTALS-Kyber) | Redes euclidianas | Intercambio de claves |
| **ML-DSA** (CRYSTALS-Dilithium) | Redes euclidianas | Firma digital |
| **SLH-DSA** (SPHINCS+) | Funciones hash | Firma digital |
| **FN-DSA** (FALCON) | NTRU Lattices | Firma digital |

> AES-256 y SHA-512 resisten mejor a Grover: en términos prácticos se compensa ampliando el tamaño de clave.

---
## C. Infraestructura de Clave Pública (PKI)

```
                     ROOT CA
               (Autoridad Raíz)
                   /      |       \
              CA1       CA2      CA3
          (Intermedias)      (Intermedias)
            /    \          /    \
        sitio1  sitio2   sitio3  sitio4
```

**Componentes PKI:**
- **CA (Certificate Authority):** entidad de confianza que emite certificados digitales
- **Certificado X.509:** documento digital que vincula una clave pública con una identidad
- **CRL (Certificate Revocation List):** lista de certificados revocados
- **OCSP:** verificación en tiempo real de la validez de un certificado
- **HSM (Hardware Security Module):** hardware especializado para proteger claves privadas

**Cadena de confianza:**
```
Tu navegador confía en ~150 Root CAs preinstaladas
Root CA firma certificado de CA Intermedia
CA Intermedia firma certificado de google.com
-> Confías en google.com porque confías en Root CA
```

````python
# ============================================================
#  DEMO: Esteganografía LSB en imágenes
# ============================================================
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

def texto_a_bits(texto: str) -> list:
    bits = []
    for char in texto:
        bits.extend([int(b) for b in format(ord(char), '08b')])
    return bits

def ocultar_mensaje_lsb(imagen: np.ndarray, mensaje: str) -> np.ndarray:
    """Oculta un mensaje en los bits menos significativos de la imagen."""
    img = imagen.copy().astype(np.uint8)
    mensaje_completo = mensaje + chr(0)  # Terminador nulo
    bits = texto_a_bits(mensaje_completo)
    flat = img.flatten()
    if len(bits) > len(flat):
        raise ValueError("Mensaje demasiado largo")
    for i, bit in enumerate(bits):
        flat[i] = (flat[i] & 0xFE) | bit  # Reemplaza LSB
    return flat.reshape(img.shape)

def extraer_mensaje_lsb(imagen: np.ndarray) -> str:
    """Extrae el mensaje oculto en los LSBs."""
    flat = imagen.flatten()
    bits = [int(p) & 1 for p in flat]
    chars = []
    for i in range(0, len(bits) - 7, 8):
        byte = int(''.join(str(b) for b in bits[i:i+8]), 2)
        if byte == 0:
            break
        chars.append(chr(byte))
    return ''.join(chars)

# Crear imagen portadora (degradado)
np.random.seed(42)
h, w = 128, 128
xx, yy = np.meshgrid(np.linspace(0, 255, w), np.linspace(0, 200, h))
imagen_original = np.stack([
    xx.astype(np.uint8),
    yy.astype(np.uint8),
    (255 - xx * 0.5).astype(np.uint8)
], axis=2)

mensaje_secreto = "Mensaje Secreto: Clave=AES256"
imagen_con_mensaje = ocultar_mensaje_lsb(imagen_original, mensaje_secreto)
mensaje_extraido = extraer_mensaje_lsb(imagen_con_mensaje)

# Visualizar
fig, axes = plt.subplots(1, 3, figsize=(13, 4))
fig.suptitle("Esteganografía LSB - El mensaje invisible", fontsize=13, fontweight='bold')

axes[0].imshow(imagen_original)
axes[0].set_title("Imagen original")
axes[0].axis('off')

axes[1].imshow(imagen_con_mensaje)
axes[1].set_title("Imagen con mensaje oculto\n(¿puedes verlo?)")
axes[1].axis('off')

diff = np.abs(imagen_original.astype(int) - imagen_con_mensaje.astype(int)) * 128
axes[2].imshow(np.clip(diff, 0, 255).astype(np.uint8))
axes[2].set_title("Diferencia (amplificada x128)")
axes[2].axis('off')

plt.tight_layout()
plt.show()

print(f"  Mensaje original : '{mensaje_secreto}'")
print(f"  Mensaje extraído : '{mensaje_extraido}'")
print(f"  Coinciden        : {mensaje_secreto == mensaje_extraido}")
print(f"\n  Capacidad utilizada: {len(mensaje_secreto)*8} bits de {h*w*3} píxeles disponibles")
print("  Cambio máximo por píxel: 1 de 255 niveles de intensidad")
````

## Disco duro cifrado (BitLocker / FileVault)
- **AES-256-XTS** cifra cada sector del disco
- La clave maestra se protege con TPM (Trusted Platform Module) + PIN
- Sin la clave: el disco es solo ruido digital

````python
# ============================================================
#  DEMO: Inspeccionar certificado TLS de un sitio web
# ============================================================
import ssl
import socket
import datetime
import json

def inspeccionar_tls(hostname: str, port: int = 443) -> dict:
    """Obtiene información del certificado TLS de un sitio."""
    ctx = ssl.create_default_context()
    try:
        with ctx.wrap_socket(socket.create_connection((hostname, port), timeout=5),
                             server_hostname=hostname) as conn:
            cert = conn.getpeercert()
            cipher = conn.cipher()
            ver = conn.version()
        return {"cert": cert, "cipher": cipher, "version": ver, "error": None}
    except Exception as e:
        return {"error": str(e)}

def mostrar_info_tls(hostname: str):
    print(f"\nInspeccionando TLS de: {hostname}")
    print("-" * 55)
    info = inspeccionar_tls(hostname)
    if info["error"]:
        print(f"  Error: {info['error']}")
        return

    cert = info["cert"]
    cipher_name, proto, key_bits = info["cipher"]

    # Sujeto del certificado
    subject = dict(x[0] for x in cert.get('subject', []))
    issuer = dict(x[0] for x in cert.get('issuer', []))

    # Fechas
    not_before = datetime.datetime.strptime(
        cert['notBefore'], '%b %d %H:%M:%S %Y %Z'
    ).replace(tzinfo=datetime.timezone.utc)
    not_after = datetime.datetime.strptime(
        cert['notAfter'], '%b %d %H:%M:%S %Y %Z'
    ).replace(tzinfo=datetime.timezone.utc)
    dias_restantes = (not_after - datetime.datetime.now(datetime.timezone.utc)).days

    print(f"  Protocolo TLS   : {info['version']}")
    print(f"  Cipher suite    : {cipher_name}")
    print(f"  Bits de clave   : {key_bits}")
    print(f"  Organización    : {subject.get('organizationName', subject.get('commonName', '?'))}")
    print(f"  CN (dominio)    : {subject.get('commonName', '?')}")
    print(f"  Emitido por     : {issuer.get('organizationName', '?')}")
    print(f"  Válido desde    : {not_before.strftime('%Y-%m-%d')}")
    print(f"  Válido hasta    : {not_after.strftime('%Y-%m-%d')} ({dias_restantes} días)")

    san = cert.get('subjectAltName', [])
    if san:
        dominios = [v for t, v in san if t == 'DNS'][:5]
        print(f"  Dominios SANs   : {', '.join(dominios)}{' ...' if len(san) > 5 else ''}")

# Inspeccionar sitios populares
for sitio in ["www.google.com", "www.wikipedia.org", "github.com"]:
    mostrar_info_tls(sitio)
````
