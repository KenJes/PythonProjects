# ALGORITMO GENÉTICO SIMPLE - METODOS DE SELECCIÓN 
**SELECCIÓN POR RANKING MÉTODO 1** 

```python
#Algoritmo Genético Simple Con Selección por Ranking forma 1:  (amax − (amax − amin) * (ranking -1) / (m - 1)) * 1 / m 
# a max = 1.2, a min = 0.2, m = 6
 
import random

individuos = [1,2,3,4,5,6] # Individuos 'X'
print("Individuos:", individuos)

def fitness(individuo):          # Función de Fitness Y = X^2
    return individuo ** 2

#Calcular el raking de cada individuo
fitness_values = [fitness(ind) for ind in individuos]

#Ordenar los individuos por su valor de fitness y asignarles un ranking
ranking = sorted(range(len(fitness_values)), key=lambda k: fitness_values[k], reverse=True)

# Probabilidad de selección por ranking
a_max = 1.2
a_min = 0.2
m = len(individuos) # m = 6
probabilidades_seleccion = [(a_max - (a_max - a_min) * (ranking.index(i)-1) / (m - 1)) * 1 / m for i in range(m)]
probablidad_acumulada = [sum(probabilidades_seleccion[:i+1]) for i in range(m)]


#Tabla de selección por ranking y probabilidades de selección
print("\nRanking de Individuos:")
print("-" * 35)
print("Individuo | Fitness | Ranking | Probabilidad | Probabilidad Acumulada")
for i in range(len(individuos)):
    print(f"{individuos[i]:>9} | {fitness_values[i]:>7} | {ranking.index(i)+1:>7} | {probabilidades_seleccion[i]:>12.6f} | {probablidad_acumulada[i]:>20.6f}")
print("-" * 35)
```

**SELECCIÓN POR RANKING MÉTODO 2**

```python
# Segundo Metodo de Selección por Ranking p = q * (1 - q)^(ranking - 1)
q = 0.5  # Probabilidad de selección del mejor individuo
probabilidades_seleccion_2 = [q * (1 - q) ** (ranking.index(i)) for i in range(m)] # Probabilidad de selección por ranking usando el segundo método
probablidad_acumulada_2 = [sum(probabilidades_seleccion_2[:i+1]) for i in range(m)] # Probabilidad acumulada para el segundo método 

#Tabla de selección por ranking y probabilidades de selección
print("\nRanking de Individuos (Método 2):")
print("-" * 35) 
print("Individuo | Fitness | Ranking | Probabilidad | Probabilidad Acumulada")
for i in range(len(individuos)):
    print(f"{individuos[i]:>9} | {fitness_values[i]:>7} | {ranking.index(i)+1:>7} | {probabilidades_seleccion_2[i]:>12.6f} | {probablidad_acumulada_2[i]:>20.6f}")
print("-" * 35)
```

**SELECCIÓN POR MÉTODO ELITISTA**

```python
# Selección de Individuos usando el metodo de selección elitista
def seleccion_elitista(individuos, fitness_values, num_selecciones):
    # Ordenar los individuos por su valor de fitness
    ranking = sorted(range(len(fitness_values)), key=lambda k: fitness_values[k], reverse=True)
    
    # Seleccionar los mejores individuos
    seleccionados = [individuos[ranking[i]] for i in range(num_selecciones)]
    
    return seleccionados

num_selecciones = 3
seleccionados_elitista = seleccion_elitista(individuos, fitness_values, num_selecciones)
print("\nIndividuos Seleccionados por el Método Elitista:")
for i, individuo in enumerate(seleccionados_elitista):
    print(f"{i+1}. {individuo}")
```

**SELECCIÓN POR TORNEO**

```python
# Selección de individuos por método de torneo. Se selecciona el individuo con el mejor fitness de un grupo de individuos seleccionados aleatoriamente. 

print("\nSelección de Individuos por Método de Torneo:")
print("-" * 50)
print("Individuos:", individuos)

print ("\nIndividuos y sus Valores de Fitness:")
for i, ind in enumerate(individuos):
    print(f"Individuo {i+1}: Valor {ind}, Fitness: {fitness_values[i]}")

def seleccion_torneo(individuos, fitness_values, num_selecciones, tamano_torneo):
    seleccionados = []
    for _ in range(num_selecciones):
        torneo = random.sample(list(zip(individuos, fitness_values)), tamano_torneo) # Seleccionar aleatoriamente un grupo de individuos para el torneo
        ganador = max(torneo, key=lambda x: x[1]) # El ganador es el individuo con el mayor valor de fitness
        seleccionados.append(ganador[0]) # Agregar el individuo ganador a la lista de seleccionados
    return seleccionados

num_selecciones_torneo = 3
tamano_torneo = 4
seleccionados_torneo = seleccion_torneo(individuos, fitness_values, num_selecciones_torneo, tamano_torneo)

# Tabñla de selección por torneo y los individuos seleccionados 

print("\n Tabla de Selección por Torneo: Individuos Seleccionados")
print("-" * 50)
print("Individuo | Fitness")
for i, individuo in enumerate(seleccionados_torneo):
    fitness_individuo = fitness_values[individuos.index(individuo)]
    print(f"{individuo:>9} | {fitness_individuo:>7}")
print("-" * 50)

```

**MÉTODO DE SELECCIÓN POR MUESTREO ESTOCÁSTICO UNIVERSAL**

```python
import random

# Selección de Individuos por Método de Selección por Muestreo Estocástico Universal
# En este método la idea es construir una ruleta con m puntos equidistantes en lugar de un solo punto de selección, lo que permite seleccionar múltiples individuos en una sola pasada. Permite seleccionar m individuos en una sola pasada, donde m es el número de individuos a seleccionar.

# Tabla de selección por muestreo estocástico universal y los individuos seleccionados 
print("\nSelección de Individuos por Método de Selección por Muestreo Estocástico Universal:")
print("-" * 50)
print("Individuo | Fitness")
if 'individuos' not in globals():
    individuos = [random.randint(1, 10) for _ in range(6)] # Generar 6 individuos aleatorios entre 1 y 10

if 'fitness_values' not in globals() or len(fitness_values) < len(individuos):
    fitness_values = [ind ** 2 for ind in individuos]

for i, ind in enumerate(individuos):
    print(f"{individuos[i]:>9} | {fitness_values[i]:>7}")
print("-" * 50)

def seleccion_muestreo_estocastico_universal(individuos, fitness_values, num_selecciones):
    total_fitness = sum(fitness_values)
    probabilidades_seleccion = [fitness / total_fitness for fitness in fitness_values]  # pt_i

    seleccionados = []
    suma = 0                                    # 1: sum ← 0
    ptr = random.uniform(0, 1)                  # 2: ptr ← rand() ∈ [0, 1]

    for i in range(len(individuos)):            # 3: para i = 1 hasta m hacer
        suma += num_selecciones * probabilidades_seleccion[i]  # 4: sum ← sum + Nt_i  (Nt_i = m · pt_i)
        while suma >= ptr:                      # 5: mientras sum ≥ ptr hacer
            seleccionados.append(individuos[i]) # 6: seleccionar individuo i
            ptr += 1                            # 7: ptr ← ptr + 1
                                                # 8: fin mientras
                                                # 9: fin para
    return seleccionados

num_selecciones_muestreo = 3
seleccionados_muestreo = seleccion_muestreo_estocastico_universal(individuos, fitness_values, num_selecciones_muestreo)
print("\nIndividuos Seleccionados por el Método de Muestreo Estocástico Universal:")
print("-" * 50) 
print("Individuo | Fitness")
for i, individuo in enumerate(seleccionados_muestreo):
    fitness_individuo = fitness_values[individuos.index(individuo)]
    print(f"{individuo:>9} | {fitness_individuo:>7}")
```

**MÉTODO DE SELECCIÓN SUSTITUCIÓN STEADY-STATE**

```python
# Método de Selección por sustitución steady-state
# En este método se selecciona un número n {1, m} de individuos los cuales serán sustituidos por n nuevos.

print("\nSelección de Individuos por Método de Selección por Sustitución Steady-State:")
print("-" * 50)
print("Individuo | Fitness")
for i, ind in enumerate(individuos):
    print(f"{individuos[i]:>9} | {fitness_values[i]:>7}")

def seleccion_steady_state(individuos, fitness_values, num_selecciones):
    seleccionados = []
    for _ in range(num_selecciones):
        individuo_seleccionado = random.choices(individuos, weights=fitness_values, k=1)[0] # Seleccionar un individuo basado en su fitness
        seleccionados.append(individuo_seleccionado)
    return seleccionados
num_selecciones_steady = 11
seleccionados_steady = seleccion_steady_state(individuos, [fitness(ind) for ind in individuos], num_selecciones_steady)
print("\nIndividuos Seleccionados por el Método de Sustitución Steady-State:")
print("-" * 50)
print("Individuo | Fitness")
for i, individuo in enumerate(seleccionados_steady):
    fitness_individuo = fitness_values[individuos.index(individuo)]
    print(f"{individuo:>9} | {fitness_individuo:>7}")

```

# OPERADORES DE CRUCE  
**CRUCE UNIFORME**

```python
# Cruce Uniforme - Cada gen en los hijos se crea copiando el correspondiente gen de un padre u otro, utilizando una máscara de cruce generada aleatoriamente.
# Donde hay un 1 en la máscara, los genes en el primer hijo se toman del primer padre y donde hay 0 los genes se toman del segundo padre.
import random

print("\nCruce Uniforme entre dos padres:")
padre1 = [1, 0, 1, 0, 1]    # Ejemplo de padre 1
padre2 = [0, 1, 0, 1, 0]    # Ejemplo de padre 2
mascara_cruce = [random.randint(0, 1) for _ in range(len(padre1))] # Generar una máscara de cruce aleatoria
hijo1 = [padre1[i] if mascara_cruce[i] == 1 else padre2[i] for i in range(len(padre1))] # Crear el primer hijo basado en la máscara de cruce
hijo2 = [padre2[i] if mascara_cruce[i] == 1 else padre1[i] for i in range(len(padre1))] # Crear el segundo hijo basado en la máscara de cruce
print("Padre 1:", padre1)
print("Padre 2:", padre2)
print("Máscara de Cruce:", mascara_cruce)
print("Hijo 1:", hijo1)
print("Hijo 2:", hijo2)
```

**CRUCE ARITMÉTICO**

```python
# Método de Cruce Aritmético - Dados 2 cromosomas seleccionados como padres para la operación de cruce se generan 2 hijos
# Ci = (bi...bilong) y Cj = (bj...bjlong) se generan 2 hijos c'i = (b'i1...b'ilong) y c'j = (b'j1...b'jlong) con b'il = a*bil + (1-a) * bjl, b'j1 = a*bjl + (1-a) * bil, para l = 1, 2,..., long, donde a es un número aleatorio entre 0 y 1.
# En resumen, cada gen del hijo es una combinación lineal de los genes correspondientes de los padres. 

print("\nCruce Aritmético entre dos padres:")
padre1 = [1,3,6,7,8]   # Ejemplo de padre 1
padre2 = [3,5,7,8,1]   # Ejemplo de padre 2
hijo1 = []
hijo2 = []
for i in range(len(padre1)):
    a = random.uniform(0, 1) # Generar un número aleatorio entre 0 y 1
    b_il = a * padre1[i] + (1 - a) * padre2[i] # Calcular el gen del primer hijo
    b_jl = a * padre2[i] + (1 - a) * padre1[i] # Calcular el gen del segundo hijo
    hijo1.append(b_il)
    hijo2.append(b_jl)
print("Padre 1:", padre1)
print("Padre 2:", padre2)
print("Hijo 1:", hijo1)
print("Hijo 2:", hijo2)

```

**CRUCE PLANO**

```python
# Método de Cruce Plano  
# Dados 2 cromosomas cti =(bti1 ...btilong) y ctj = (btj1 ...btjlong) seleccionados como padres para la operación de cruce, se genera el cromosoma ctk =(btk1 ...btklong) con btkl generado aleatoria y uniformemente en el intervalo [btil, btjl], para l =1, . . . , long.
# En resumen, el gen del hijo se genera aleatoriamente dentro del intervalo definido por los genes correspondientes de los padres.
print("\nCruce Plano entre dos padres:")
padre1 = [1,3,6,7,8]   # Ejemplo de padre 1
padre2 = [3,5,7,8,1]   # Ejemplo de padre 2
hijo = []
for i in range(len(padre1)):
    b_til = min(padre1[i], padre2[i]) # Calcular el límite inferior del intervalo
    b_tjl = max(padre1[i], padre2[i]) # Calcular el límite superior del intervalo
    b_tkl = random.uniform(b_til, b_tjl) # Generar un número aleatorio en el intervalo [btil, btjl]
    hijo.append(b_tkl)  # Agregar el gen del hijo a la lista del hijo
print("Padre 1:", padre1)
print("Padre 2:", padre2)
print("Hijo:", hijo)
```

**Cruce BLX-α** 

```python
# Método de Cruce BLX-α
# Dados 2 cromosomas cti =(bti1 ...btilong) y ctj = (btj1 ...btjlong)
# se genera el cromosoma ctk =(btk1 ...btklong) con btkl generado aleatoria y uniformemente en el intervalo
# [mintkj - Iα, maxtkl + Iα], con mintkl = min(btil, btjl), maxtkl = max(btil, btjl), I = maxtkl - mintkl, para l = 1, 2,..., long.
# El cruce plano es un caso particular del cruce BLX-α con α = 0, ya que el intervalo de generación de los genes del hijo se reduce al intervalo definido por los genes correspondientes de los padres. 
print("\nCruce BLX-α entre dos padres:")
padre1 = [1,3,6,7,8]   # Ejemplo de padre 1
padre2 = [3,5,7,8,1]   # Ejemplo de padre 2
alpha = 0.0 # Valor de α

hijo_blx = []
for i in range(len(padre1)):
    b_til = min(padre1[i], padre2[i]) # Calcular el límite inferior del intervalo
    b_tjl = max(padre1[i], padre2[i]) # Calcular el límite superior del intervalo
    I = b_tjl - b_til # Calcular la amplitud del intervalo
    b_tkl = random.uniform(b_til - I * alpha, b_tjl + I * alpha) # Generar un número aleatorio en el intervalo [mintkl - Iα, maxtkl + Iα]
    hijo_blx.append(b_tkl)  # Agregar el gen del hijo a la lista del hijo

print("Padre 1:", padre1)
print("Padre 2:", padre2)
print("Hijo BLX-α:", hijo_blx)
```

**Cruce PMX, OX y CX**

 Usados cuando se utilizan permutaciones como represencación

```python
# Método de Cruce PMX (Partially Mapped Crossover) - Este método se utiliza para problemas de permutación, como el problema del viajante de comercio. Se selecciona aleatoriamente un segmento de genes de un padre y se copia al hijo. Luego, se mapean los genes restantes del otro padre para llenar los espacios vacíos en el hijo, asegurando que no haya duplicados.
print("\nCruce PMX entre dos padres:")
padre1 = [2, 4, 5, 6, 9]   # Ejemplo de padre 1
padre2 = [3, 5, 7, 8, 1]   # Ejemplo de padre 2
hijo_pmx = [None] * len(padre1) # Inicializar el hijo
# Seleccionar aleatoriamente un segmento de genes del primer padre
inicio_segmento = random.randint(0, len(padre1) - 2)
fin_segmento = random.randint(inicio_segmento + 1, len(padre1) - 1)
# Copiar el segmento seleccionado del primer padre al hijo
for i in range(inicio_segmento, fin_segmento + 1):
    hijo_pmx[i] = padre1[i]
# Mapear los genes restantes del segundo padre para llenar los espacios vacíos en el hijo
for i in range(len(padre2)):
    if padre2[i] not in hijo_pmx:
        for j in range(len(padre1)):
            if padre1[j] == padre2[i]:
                hijo_pmx[j] = padre2[i]
                break
print("Padre 1:", padre1)
print("Padre 2:", padre2)
print("Hijo PMX:", hijo_pmx)
```

```python
# Método de Cruce OX (Order Crossover) - Este método también se utiliza para problemas de permutación. Se selecciona aleatoriamente un segmento de genes de un padre y se copia al hijo. Luego, se rellenan los espacios vacíos en el hijo con los genes del otro padre, manteniendo el orden relativo de los genes.
print("\nCruce OX entre dos padres:")
padre1 = [2, 4, 5, 6, 9]   # Ejemplo de padre 1
padre2 = [3, 5, 7, 8, 1]   # Ejemplo de padre 2
hijo_ox = [None] * len(padre1) # Inicializar el hijo
# Seleccionar aleatoriamente un segmento de genes del primer padre
inicio_segmento = random.randint(0, len(padre1) - 2)
fin_segmento = random.randint(inicio_segmento + 1, len(padre1) - 1)
# Copiar el segmento seleccionado del primer padre al hijo
for i in range(inicio_segmento, fin_segmento + 1):
    hijo_ox[i] = padre1[i]
# Rellenar los espacios vacíos en el hijo con los genes del segundo padre, manteniendo el orden relativo
posicion_hijo = 0
for i in range(len(padre2)):
    if padre2[i] not in hijo_ox:
        while posicion_hijo < len(hijo_ox) and hijo_ox[posicion_hijo] is not None:
            posicion_hijo += 1
        if posicion_hijo >= len(hijo_ox):
            break
            posicion_hijo += 1
        hijo_ox[posicion_hijo] = padre2[i]
print("Padre 1:", padre1)
print("Padre 2:", padre2)
print("Hijo OX:", hijo_ox)
```

```python
# Método de Cruce CX (Cycle Crossover) - Este método también se utiliza para problemas de permutación. Se identifican ciclos entre los dos padres y se copian los genes de un padre al hijo para los genes que pertenecen a un ciclo, mientras que los genes del otro padre se copian para los genes que no pertenecen a ese ciclo.
print("\nCruce CX entre dos padres:")
padre1 = [2, 4, 5, 6, 9]   # Ejemplo de padre 1
padre2 = [3, 5, 7, 8, 1]   # Ejemplo de padre 2
hijo_cx = [None] * len(padre1) # Inicializar el hijo
# Identificar ciclos entre los dos padres
ciclo = []
visitados = [False] * len(padre1)   # Inicializar una lista para marcar los genes visitados
for i in range(len(padre1)):
    if not visitados[i]:
        ciclo_actual = []
        j = i
        while not visitados[j]:
            visitados[j] = True
            ciclo_actual.append(j)
            try:
                j = padre1.index(padre2[j])  # Encontrar el índice del gen correspondiente en el otro padre
            except ValueError:
                break  # Evita error si los padres no son permutaciones compatibles
        ciclo.append(ciclo_actual)
# Copiar los genes de un padre al hijo para los genes que pertenecen a un ciclo, mientras que los genes del otro padre se copian para los genes que no pertenecen a ese ciclo
for i in range(len(ciclo)):
    if i % 2 == 0: # Para los ciclos pares, copiar los genes del primer padre
        for j in ciclo[i]:
            hijo_cx[j] = padre1[j]
    else: # Para los ciclos impares, copiar los genes del segundo padre
        for j in ciclo[i]:
            hijo_cx[j] = padre2[j]
print("Padre 1:", padre1)
print("Padre 2:", padre2)
print("Hijo CX:", hijo_cx)
```

# Operadores de Mutación
**Mutación Uniforme**

```python
# Método de Mutación Uniforme
# El operador de mutación uniforme modifica un gen especifico por un valor generado aleatoria y uniformemente en el dominio de la variable
print("\nMutación Uniforme en un individuo:")
individuo = [1, 2, 3, 4, 5] # Ejemplo de individuo
print ("Individuo antes de la mutación:", individuo)
gen_mutar = random.randint(0, len(individuo) - 1) # Seleccionar aleatoriamente un gen para mutar
valor_mutacion = random.uniform(0, 10) # Generar un valor de mutación aleatorio entre 0 y 10
individuo[gen_mutar] = valor_mutacion # Modificar el gen seleccionado con el valor de mutación
print("Individuo después de la mutación:", individuo)
```

**Mutación no uniforme** 

```python
# Método de Mutación no uniforme
# Dado el cromosoma cti = (bti1 ...btilong) y supuesto que el gen btil ha sido seleccionado para mutar
# el resultado es c'ti1 = (bti1...b'til...btilong) donde b'til es para representaciones reales

print("\nMutación no uniforme en un individuo:")
individuo = [1, 2, 3, 4, 5] # Ejemplo de individuo
print ("Individuo antes de la mutación:", individuo)
gen_mutar = random.randint(0, len(individuo) - 1) # Seleccionar aleatoriamente un gen para mutar
valor_mutacion = random.uniform(0, 10) # Generar un valor de mutación aleatorio entre 0 y 10
individuo[gen_mutar] = valor_mutacion # Modificar el gen seleccionado con el valor de mutación
print("Individuo después de la mutación:", individuo)
```

**Método de Mutación por Intercambio**

```python
# Método de Mutación por Intercambio
# Se intercambian 2 elementos elegidos aleatoriamente de una permutación
print("\nMutación por Intercambio en un individuo:")
individuo = [14, 7, 33, 94, 5] # Ejemplo de individuo
print ("Individuo antes de la mutación:", individuo)
gen1_mutar = random.randint(0, len(individuo) - 1) # Seleccionar aleatoriamente el primer gen para mutar
gen2_mutar = random.randint(0, len(individuo) - 1) # Seleccionar aleatoriamente el segundo gen para mutar
# Asegurarse de que los dos genes seleccionados sean diferentes para evitar intercambiar el mismo gen
while gen1_mutar == gen2_mutar:
    gen2_mutar = random.randint(0, len(individuo) - 1)
# Intercambiar los genes seleccionados
individuo[gen1_mutar], individuo[gen2_mutar] = individuo[gen2_mutar], individuo[gen1_mutar]
print("Individuo después de la mutación por intercambio:", individuo)   
```
