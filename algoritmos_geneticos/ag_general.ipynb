# Algoritmo Genético General con Hiperparametros Ajustables Y diferentes metodos de selección, cruza y mutación 

**Se definen los hiperparámetros del Algoritmo Genético**

```python
#Algoritmo Genético con hiperparametros ajustables y diferentes funciones de aptitud

# Parametros del libro IEEE TRANSACTIONS ON EVOLUTIONARY COMPUTATION, VOL. 8, NO. 4, AUGUST 2004 
# RECOMMENDATIONS FOR BASIC PARAMETER SETTINGS Configuración:'De Jong' 

poblacion = 75
generaciones = 10  
tasa_mutacion = 0.001
tasa_cruce = 0.6
bits_por_individuo = 20
```

```python
# Paso 1 - Inicialización de la población
from random import random

def inicializar_poblacion(poblacion, bits_por_individuo):
    return [[1 if random() >= 0.5 else 0 for _ in range(bits_por_individuo)] for _ in range(poblacion)]
# Se muestra una tabla con la población inicial generada con los parámetros establecidos
poblacion_inicial = inicializar_poblacion(poblacion, bits_por_individuo)
print("Población inicial:")
for i, individuo in enumerate(poblacion_inicial):
    print(f"Individuo {i+1}: {individuo}")
```

```python
# Paso 2 - Evaluación de la función de fitness
def evaluar_fitness(individuo):
    # Función de fitness: f(x) = Σ i = 1 xi^2 y xi >= -5.12 y xi <= 5.12 es decir que es la suma ponderada de los cuadrados de 3 variables con un rango de -5.12 a 5.12
    return sum(x**2 for x in individuo)

# Se muestra una tabla con los individuos y su fitness correspondiente
poblacion_inicial = inicializar_poblacion(poblacion, bits_por_individuo)
fitness_inicial = [evaluar_fitness(ind) for ind in poblacion_inicial]
print("\nIndividuos y su fitness correspondiente:")
for i, (ind, fit) in enumerate(zip(poblacion_inicial, fitness_inicial)):
    print(f"Individuo {i+1}: {ind}, Fitness: {fit}")
```

```python
# Paso 3 - Selección de los padres con diferentes métodos de selección

def ruleta(poblacion, fitness):
    total_fitness = sum(fitness)
    seleccion = random() * total_fitness
    acumulado = 0
    for i, f in enumerate(fitness):
        acumulado += f
        if acumulado >= seleccion:
            return poblacion[i]
        
def torneo(poblacion, fitness, k=3):
    seleccionados = random.sample(list(zip(poblacion, fitness)), k)
    return max(seleccionados, key=lambda x: x[1])[0]

def muestreo_estocastico(poblacion, fitness):
    total_fitness = sum(fitness)
    seleccion = random() * total_fitness
    acumulado = 0
    for i, f in enumerate(fitness):
        acumulado += f
        if acumulado >= seleccion:
            return poblacion[i]
        
def ranking(poblacion, fitness): # Metodo de selección por ranking con amax y amin y probabilidad de selección proporcional al ranking y probabilidad acumulada
    amax = max(fitness)
    amin = min(fitness)
    ranking = [(i, (f - amin) / (amax - amin)) for i, f in enumerate(fitness)]
    total_ranking = sum(r[1] for r in ranking)
    seleccion = random() * total_ranking
    acumulado = 0
    for i, r in ranking:
        acumulado += r
        if acumulado >= seleccion:
            return poblacion[i]

def elitista(poblacion, fitness):
    mejor_individuo = max(zip(poblacion, fitness), key=lambda x: x[1])[0]
    return mejor_individuo

print ("Elija el método de selección de padres: ")
print("1. Ruleta")
print("2. Torneo")
print("3. Muestreo Estocástico")
print("4. Ranking")
print("5. Elitista")

opcion = int(input("Ingrese el número correspondiente al método de selección: "))
if opcion == 1:
    metodo_seleccion = ruleta
elif opcion == 2:    
    metodo_seleccion = torneo
elif opcion == 3:    
    metodo_seleccion = muestreo_estocastico
elif opcion == 4:    
    metodo_seleccion = ranking
elif opcion == 5:    
    metodo_seleccion = elitista
else:    
    print("Opción no válida, se utilizará el método de selección por ruleta por defecto.")
    metodo_seleccion = ruleta
    
print(f"\nMétodo de selección elegido: {metodo_seleccion.__name__}")
print("\nIndividuos seleccionados para reproducción:")
for i in range(5):  # Seleccionamos 5 individuos para mostrar
    padre = metodo_seleccion(poblacion_inicial, fitness_inicial)
    print(f"Padre {i+1}: {padre}, Fitness: {evaluar_fitness(padre)}")
```

```python
import random as rnd

# Paso 4 - Cruce de los padres seleccionados con diferentes métodos de cruce 

def cruce_uniforme(padre1, padre2):
    hijo = []
    for g1, g2 in zip(padre1, padre2):
        hijo.append(g1 if random() < 0.5 else g2)
    return hijo

def cruce_aritmetico(padre1, padre2):
    return [(g1 + g2) / 2 for g1, g2 in zip(padre1, padre2)]    

def cruce_plano(padre1, padre2):
    punto_cruce = random.randint(1, len(padre1) - 1)
    hijo = padre1[:punto_cruce] + padre2[punto_cruce:]
    return hijo

def cruce_BLX(padre1, padre2):
    alpha = random.random()
    hijo = [alpha * g1 + (1 - alpha) * g2 for g1, g2 in zip(padre1, padre2)]
    return hijo

def cruce_PMX(padre1, padre2):
    punto_cruce1 = random.randint(0, len(padre1) - 2)
    punto_cruce2 = random.randint(punto_cruce1 + 1, len(padre1) - 1)
    hijo = [None] * len(padre1)
    hijo[punto_cruce1:punto_cruce2] = padre1[punto_cruce1:punto_cruce2]
    for g in padre2:
        if g not in hijo:
            for i in range(len(hijo)):
                if hijo[i] is None:
                    hijo[i] = g
                    break
    return hijo

def cruce_OX(padre1, padre2):
    punto_cruce1 = random.randint(0, len(padre1) - 2)
    punto_cruce2 = random.randint(punto_cruce1 + 1, len(padre1) - 1)
    hijo = [None] * len(padre1)
    hijo[punto_cruce1:punto_cruce2] = padre1[punto_cruce1:punto_cruce2]
    for g in padre2:
        if g not in hijo:
            for i in range(len(hijo)):
                if hijo[i] is None:
                    hijo[i] = g
                    break
    return hijo

def cruce_CX(padre1, padre2):
    hijo = [None] * len(padre1)
    for i in range(len(padre1)):
        if padre1[i] == padre2[i]:
            hijo[i] = padre1[i]
    for i in range(len(padre1)):
        if hijo[i] is None:
            for j in range(len(padre1)):
                if padre2[j] == padre1[i] and hijo[j] is None:
                    hijo[j] = padre1[i]
                    break
    return hijo

print ("\nElija el método de cruce de padres: ")
print("1. Cruce Uniforme")
print("2. Cruce Aritmético")
print("3. Cruce Plano")
print("4. Cruce BLX")
print("5. Cruce PMX")
print("6. Cruce OX")
print("7. Cruce CX")

opcion_cruce = int(input("Ingrese el número correspondiente al método de cruce: "))
if opcion_cruce == 1:
    metodo_cruce = cruce_uniforme
elif opcion_cruce == 2:    
    metodo_cruce = cruce_aritmetico
elif opcion_cruce == 3:    
    metodo_cruce = cruce_plano
elif opcion_cruce == 4:    
    metodo_cruce = cruce_BLX
elif opcion_cruce == 5:
    metodo_cruce = cruce_PMX
elif opcion_cruce == 6:
    metodo_cruce = cruce_OX
elif opcion_cruce == 7:
    metodo_cruce = cruce_CX
else:    
    print("Opción no válida, se utilizará el método de cruce uniforme por defecto.")
    metodo_cruce = cruce_uniforme
print(f"\nMétodo de cruce elegido: {metodo_cruce.__name__}")

# Se muestra una tabla con los padres seleccionados y sus hijos generados con el método de cruce elegido
print("\nPadres seleccionados y sus hijos generados:")
for i in range(5):  # Seleccionamos 5 pares de padres para mostrar
    padre1 = metodo_seleccion(poblacion_inicial, fitness_inicial)
    padre2 = metodo_seleccion(poblacion_inicial, fitness_inicial)
    def cruce_uniforme(padre1, padre2):
        hijo = []
        for g1, g2 in zip(padre1, padre2):
            hijo.append(g1 if rnd.random() < 0.5 else g2)
        return hijo

    def cruce_aritmetico(padre1, padre2):
        return [(g1 + g2) / 2 for g1, g2 in zip(padre1, padre2)]    

    def cruce_plano(padre1, padre2):
        punto_cruce = rnd.randint(1, len(padre1) - 1)
        hijo = padre1[:punto_cruce] + padre2[punto_cruce:]
        return hijo

    def cruce_BLX(padre1, padre2):
        alpha = rnd.random()
        hijo = [alpha * g1 + (1 - alpha) * g2 for g1, g2 in zip(padre1, padre2)]
        return hijo

    def cruce_PMX(padre1, padre2):
        punto_cruce1 = rnd.randint(0, len(padre1) - 2)
        punto_cruce2 = rnd.randint(punto_cruce1 + 1, len(padre1) - 1)
        hijo = [None] * len(padre1)
        hijo[punto_cruce1:punto_cruce2] = padre1[punto_cruce1:punto_cruce2]
        for g in padre2:
            if g not in hijo:
                for i in range(len(hijo)):
                    if hijo[i] is None:
                        hijo[i] = g
                        break
        return hijo

    def cruce_OX(padre1, padre2):
        punto_cruce1 = rnd.randint(0, len(padre1) - 2)
        punto_cruce2 = rnd.randint(punto_cruce1 + 1, len(padre1) - 1)
        hijo = [None] * len(padre1)
        hijo[punto_cruce1:punto_cruce2] = padre1[punto_cruce1:punto_cruce2]
        for g in padre2:
            if g not in hijo:
                for i in range(len(hijo)):
                    if hijo[i] is None:
                        hijo[i] = g
                        break
        return hijo
    print(f"Padre 1: {padre1}, Fitness: {evaluar_fitness(padre1)}")
    print(f"Padre 2: {padre2}, Fitness: {evaluar_fitness(padre2)}")
    metodos_cruce = {
        1: cruce_uniforme,
        2: cruce_aritmetico,
        3: cruce_plano,
        4: cruce_BLX,
        5: cruce_PMX,
        6: cruce_OX,
        7: cruce_CX,
    }
    hijo = metodos_cruce.get(opcion_cruce, cruce_uniforme)(padre1, padre2)
    print(f"Hijo {i + 1}: {hijo}, Fitness: {evaluar_fitness(hijo)}\n")

```
